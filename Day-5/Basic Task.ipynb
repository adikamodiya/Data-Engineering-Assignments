# Question 1

Load the messy e-commerce dataset and identify columns containing nulls and duplicate rows using .filter(), .distinct(), and .dropDuplicates(). 

```python
df = spark.read.option('header', 'true').option('inferSchema', 'true').csv('/Volumes/cyntexa_dev/sales/my_volume/ecommerce.csv')
```

```python
df.columns
```

```python
from pyspark.sql.functions import *

# 1. Identify columns containing NULLs
cols = []

for c in df.columns:
    if df.filter(col(c).isNull()).count() > 0:
        cols.append(c)

print(cols)


# 2. Check duplicate rows
total_rows = df.count()
distinct_rows = df.distinct().count()

print("Total rows:", total_rows)
print("Distinct rows:", distinct_rows)
print("Duplicate rows:", total_rows - distinct_rows)


# 3. Remove duplicate rows
df_distinct = df.dropDuplicates()


print("Rows after removing duplicates:", df_distinct.count())
```
