# Question 2
Use COPY INTO to incrementally load 2 batches of files into a bronze table, confirming COPY INTO doesn't reprocess the first batch. 

```sql
create volume if not exists cyntexa_dev.dev.raw
```

```sql
create table if not exists cyntexa_dev.dev.increamental_table;
```

```sql
COPY INTO cyntexa_dev.dev.increamental_table
FROM (
    SELECT
        *,
        _metadata.file_name AS source_file_name,
        _metadata.file_path AS source_file_path,
        _metadata.file_modification_time AS source_file_modification_time,
        current_timestamp() AS ingestion_timestamp
    FROM '/Volumes/cyntexa_dev/dev/raw/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true');
```

```sql
select source_file_name from  cyntexa_dev.dev.increamental_table;
```
