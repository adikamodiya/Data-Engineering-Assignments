**Warhouse** -  It is the storage space where  all the structured data is stored from the different databases
    Example data - Tables. 


**Data Lake** -    It is the storage space where all the unstructure , semi-structured  data is stored Example data - Jsons, xml (semi structured),  pdg, png ,jpg (unstructured) . 


**Data Lakehouse** -  It is the combination of warehouses and data lake , where any type of data can be stored  whether it is structured (tables) , semi structured (csv ,jsons , xml) or unstructured(pdf , jpb ,png).
