# Question 1

Install and authenticate the Databricks CLI using OAuth U2M against your workspace. 


1. Check Databricks CLI installation
   ```databricks -v```

2. Login to existing Databricks Workspace using OAuth U2M
   ```databricks auth login --host <workspace-url> --profile DEFAULT```

   - Browser will open automatically.
   - Login to Databricks using your account.
   - After successful authentication, return to the terminal.
   - Terminal will show a success message:
     `Profile DEFAULT was successfully saved`

3. Check authentication profiles
   ```databricks auth profiles```

4. Verify current logged-in Databricks user
   ```databricks current-user me```

5. Check Workspace access / list Workspace contents
   ```databricks workspace list /```
