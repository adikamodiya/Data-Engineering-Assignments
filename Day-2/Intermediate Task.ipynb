```sql
create or replace view cyntexa_dev.sales.total_spent_per_customer AS   
(
SELECT
    c.customer_id,
    CONCAT(c.first_name, ' ', c.last_name) AS customer_name,
    SUM(o.total_amount) AS total_spent
FROM cyntexa_dev.sales.orders_views o
LEFT JOIN cyntexa_dev.sales.customers c
    ON o.customer_id = c.customer_id
GROUP BY
    c.customer_id,
    c.first_name,
    c.last_name
)
;
```
