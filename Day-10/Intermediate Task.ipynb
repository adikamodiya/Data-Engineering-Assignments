# Question 5

Set up a Genie Agent scoped to your sales tables: write at least 2 instructions and 2 sample queries
to steer its behavior.

Query given to agent - What is the distribution of sales by region?


code given by agent - 
```SELECT
  `region`,
  SUM(`sale_amount`) AS total_sales
FROM
  `dev`.`silver`.`sales_cleaned`
WHERE
  `region` IS NOT NULL
GROUP BY
  `region`
ORDER BY
  total_sales DESC;
  ```


  ![image_1789375892458.png](./image_1789375892458.png "image_1789375892458.png")

# Question 6

Ask your Genie Agent a question it initially answers incorrectly or vaguely, then improve its
instructions/sample queries/trusted assets until it answers correctly — document the before and
after.
