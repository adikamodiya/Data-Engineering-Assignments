# Genie Agent Curation Checklist – Cyntexa

Before allowing a Genie Agent to answer executive-facing questions, the underlying Unity Catalog assets should pass the following checks.

## 1. Table-Level Metadata

* [ ] Every table has a clear table description.
* [ ] The description explains the business purpose of the table.
* [ ] The table represents trusted and validated data.
* [ ] Raw or unvalidated tables are not used for executive reporting.

## 2. Column-Level Metadata

* [ ] Important columns have meaningful column comments.
* [ ] Column comments explain the business meaning of each field.
* [ ] Date, ID, amount, and status columns are clearly defined.
* [ ] Sensitive/PII columns are identified appropriately.

## 3. Business Metrics

* [ ] Important metrics have clear business definitions.
* [ ] Revenue, AOV, growth, customer lifetime value, etc. have documented calculation logic.
* [ ] Metric definitions are consistent across tables and reports.

## 4. Data Relationships

* [ ] Relationships between important tables are documented.
* [ ] Primary and foreign key relationships are clear where applicable.
* [ ] Genie can understand how tables should be joined.

## 5. Data Quality

* [ ] Important columns have appropriate data types.
* [ ] Null and duplicate issues have been addressed.
* [ ] Data freshness/refresh frequency is known.
* [ ] The data has been validated before being exposed to executives.

## 6. Naming Standards

* [ ] Table names are meaningful.
* [ ] Column names are meaningful and business-friendly.
* [ ] Abbreviations are avoided or documented.

## 7. Genie Validation

* [ ] Common executive questions have been tested.
* [ ] Genie answers have been compared with known/verified results.
* [ ] Incorrect or ambiguous answers are investigated before production use.
* [ ] Trusted tables and views are used as Genie data sources.

## Final Approval

A Genie Agent should only be considered ready for executive-facing questions when the required Unity Catalog metadata, business definitions, data quality checks, and Genie answer validation have been completed.

# Serverless SQL Warehouse for Fast Ad-Hoc Dashboard Queries

## Overview

A serverless SQL Warehouse is well suited for ad-hoc dashboard queries because it combines optimized query execution, efficient data access, and intelligent workload management.

Three important capabilities that contribute to this performance are:

- Photon
- Predictive I/O
- Intelligent Workload Management

Together, these capabilities help serverless warehouses provide fast and responsive dashboard experiences.

---

## 1. Photon

Photon is Databricks' high-performance query engine designed to accelerate SQL and analytical workloads.

It improves query execution for operations such as:

- Table scans
- Joins
- Aggregations
- Filtering
- Sorting

For example, when a dashboard query calculates total sales by product category, Photon can execute the SQL operations efficiently and reduce query execution time.

**Role in dashboard performance:**  
Photon mainly improves the **speed of query execution**.

---

## 2. Predictive I/O

Predictive I/O improves how data is read during query execution.

Instead of treating every data read equally, it uses intelligent techniques to optimize data access and reduce unnecessary I/O operations.

This is particularly useful for dashboards because users may frequently apply different filters and request different subsets of data.

For example, if a dashboard query only needs sales data for a particular year and region, efficient data access can reduce the amount of unnecessary data that needs to be processed.

**Role in dashboard performance:**  
Predictive I/O mainly improves the **efficiency of data reading**.

---

## 3. Intelligent Workload Management

Intelligent Workload Management helps manage compute resources based on the workload running on the warehouse.

Dashboard environments can have many users running queries at the same time. Some queries may be simple while others may involve large joins or aggregations.

Intelligent workload management helps the warehouse handle these workloads efficiently and maintain responsiveness under concurrent usage.

**Role in dashboard performance:**  
Intelligent Workload Management mainly improves **resource utilization and query responsiveness under concurrent workloads**.

---

## Comparison

| Capability | Main Role | Benefit for Dashboards |
|---|---|---|
| **Photon** | Fast query execution | Faster joins, aggregations, scans, and SQL operations |
| **Predictive I/O** | Efficient data access | Reduces unnecessary data reading and improves I/O efficiency |
| **Intelligent Workload Management** | Intelligent resource management | Helps maintain performance when multiple users run queries |

These capabilities work together:

**User Query → Predictive I/O → Efficient Data Access → Photon → Fast Query Execution → Intelligent Workload Management → Responsive Dashboard**

---

## Why Serverless is Suitable for Ad-Hoc Dashboards

Ad-hoc dashboard workloads are usually unpredictable. Users can run different queries, apply different filters, and access dashboards simultaneously.

Serverless is a strong choice for this use case because it provides optimized query execution and workload management without requiring users to manage the underlying infrastructure.

The combination of Photon, Predictive I/O, and Intelligent Workload Management helps provide:

- Fast query execution
- Efficient data access
- Better handling of concurrent workloads
- Reduced infrastructure management
- A more responsive dashboard experience

---

## Serverless vs Classic/Pro Warehouses

| Aspect | Serverless | Classic/Pro |
|---|---|---|
| Infrastructure management | More managed by Databricks | More infrastructure/configuration responsibility |
| Query optimization | Photon and serverless optimizations | Can also provide high performance, depending on configuration |
| Data access | Predictive I/O capabilities | More dependent on configuration/workload |
| Workload management | Intelligent workload management | More dependent on configured resources |
| Ad-hoc dashboard workload | **Very suitable** | Suitable, but may require more configuration/tuning |
| Operational overhead | Lower | Higher |

It is important to note that Classic/Pro warehouses are not necessarily slow. They can also provide good performance when properly configured. However, for unpredictable and concurrent dashboard workloads, serverless provides a more managed and optimized experience.

---

## Conclusion

For ad-hoc dashboard queries, Serverless SQL Warehouse is a strong choice because Photon accelerates query execution, Predictive I/O improves data access efficiency, and Intelligent Workload Management helps handle concurrent workloads effectively.

Therefore, serverless is preferred over Classic/Pro for this use case because it combines **performance, workload management, and lower infrastructure-management overhead**, making it suitable for fast and responsive analytical dashboards.
