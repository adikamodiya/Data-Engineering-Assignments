```python
from pyspark.sql.types import *

sch = StructType([
    StructField('name', StringType(), True),
    StructField('age', IntegerType(), True),
    StructField('city', StringType(), True),
    StructField('salary', DoubleType(), True),
    StructField('is_active', BooleanType(), False)
])

df_explicit = spark.read.csv(
    '/Volumes/cyntexa_dev/sales/day_4/data2.csv',
    header=True,
    schema=sch
)

# df.printSchema()

df_explicit.printSchema()
```

```python
from pyspark.sql.functions import *

df = spark.read.csv(
    '/Volumes/cyntexa_dev/sales/day_4/data2.csv',
    header=True,
    inferSchema=True
)

filtered = df.filter("age > 20 AND is_active = false")
```

> When a transformation (e.g., filter) is applied to a DataFrame, it is not executed immediately because Spark follows lazy evaluation. The transformation is executed only when an action, such as count() or show(), is called.
